```python

import os
import json
import ast
import operator

from groq import Groq
from ddgs import DDGS

# Initialize Groq client
client = Groq(
    api_key=os.environ["GROQ_API_KEY"]
)

# Select the LLM model
MODEL = "openai/gpt-oss-120b"

print("LLM initialized successfully!")
```

```python

def calculator(expression):

    # Allowed mathematical operators
    allowed_operators = {
        ast.Add: operator.add,
        ast.Sub: operator.sub,
        ast.Mult: operator.mul,
        ast.Div: operator.truediv,
        ast.Pow: operator.pow,
        ast.Mod: operator.mod,
        ast.USub: operator.neg
    }

    # Evaluate mathematical expression safely
    def evaluate(node):

        if isinstance(node, ast.Constant):
            if type(node.value) in (int, float):
                return node.value

        elif isinstance(node, ast.BinOp):

            op = type(node.op)

            if op not in allowed_operators:
                raise ValueError("Operator not allowed")

            return allowed_operators[op](
                evaluate(node.left),
                evaluate(node.right)
            )

        elif isinstance(node, ast.UnaryOp):

            op = type(node.op)

            if op not in allowed_operators:
                raise ValueError("Operator not allowed")

            return allowed_operators[op](
                evaluate(node.operand)
            )

        raise ValueError("Invalid expression")

    try:
        tree = ast.parse(expression, mode="eval")
        result = evaluate(tree.body)

        return str(result)

    except Exception as e:
        return "Error: " + str(e)


# Test calculator
print(calculator("250 * 48 + 500"))
```

```python

def search_tool(query):

    try:
        # Search the internet
        results = DDGS().text(
            query,
            max_results=5
        )

        if not results:
            return "No search results found."

        output = []

        # Extract search results
        for item in results:

            output.append(
                "Title: " + item.get("title", "") +
                "\nDescription: " + item.get("body", "") +
                "\nURL: " + item.get("href", "")
            )

        return "\n\n".join(output)

    except Exception as e:
        return "Search error: " + str(e)


# Test search tool
print(search_tool("Who invented the World Wide Web?"))
```

```python

SYSTEM_PROMPT = """
You are a ReAct-based AI assistant.

You have access to two tools:

1. calculator:
   Use this tool for mathematical calculations.
   Input: A mathematical expression.

2. search:
   Use this tool for factual and current information.
   Input: A search query.

Follow this process:

Thought -> Action -> Observation -> Final Answer.

Return ONLY valid JSON in this format:

{
    "thought": "Brief explanation of the next decision",
    "action": "search, calculator, or final",
    "action_input": "Search query or mathematical expression",
    "final_answer": "Final response if action is final"
}

Rules:
- Use search for factual information.
- Use calculator for calculations.
- Use the observation returned by the tool.
- You may call multiple tools if required.
- Never invent tool observations.
- When you have enough information, choose final.
- Keep the thought field brief.
- Do not reveal private chain-of-thought reasoning.
"""

print("ReAct prompt defined successfully!")
```

```python

def react_agent(question):

    # Store conversation history
    messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT
        },
        {
            "role": "user",
            "content": question
        }
    ]

    # Maximum number of iterations
    max_iterations = 5

    for step in range(max_iterations):

        print("\n========== STEP", step + 1, "==========")

        # Ask LLM to decide the next action
        response = client.chat.completions.create(
            model=MODEL,
            messages=messages,
            temperature=0.2,
            response_format={"type": "json_object"}
        )

        # Convert JSON response into Python dictionary
        result = json.loads(
            response.choices[0].message.content
        )

        thought = result.get("thought", "")
        action = result.get("action", "").lower()
        action_input = result.get("action_input", "")

        print("\nTHOUGHT:", thought)

        # Check if final answer is ready
        if action == "final":

            answer = result.get("final_answer", "")

            print("\nFINAL ANSWER:")
            print(answer)

            return answer

        # Execute calculator tool
        elif action == "calculator":

            print("\nACTION: Calculator")
            print("INPUT:", action_input)

            observation = calculator(action_input)

        # Execute search tool
        elif action == "search":

            print("\nACTION: Search")
            print("INPUT:", action_input)

            observation = search_tool(action_input)

        else:

            observation = (
                "Invalid action. Choose search, calculator or final."
            )

        # Display tool output
        print("\nOBSERVATION:")
        print(observation)

        # Add LLM decision to conversation history
        messages.append({
            "role": "assistant",
            "content": json.dumps(result)
        })

        # Send observation back to LLM
        messages.append({
            "role": "user",
            "content": (
                "TOOL OBSERVATION:\n"
                + observation
                + "\nUse this observation to continue. "
                "Select another tool or provide the final answer."
            )
        })

    return "Maximum iterations reached. Please try again."


print("ReAct Agent created successfully!")
```

```python

print("===================================")
print("       ReAct AI ASSISTANT")
print("===================================")

while True:

    question = input(
        "\nEnter your question (or type exit): "
    )

    if question.lower() == "exit":
        print("Thank you for using ReAct Assistant!")
        break

    react_agent(question)
```

```python
import gradio as gr
from google import genai
from google.genai import types
from PIL import Image
from io import BytesIO
from getpass import getpass

# Enter your API key securely
API_KEY = getpass("Enter Gemini API Key: ")

# Initialize Gemini client
client = genai.Client(api_key=API_KEY)

MODEL = "gemini-2.5-flash"

print("Gemini Vision Model Initialized Successfully!")
```

```python
def generate_description(image):

    if image is None:
        return "Please upload an image first!"

    try:
        # Convert image into bytes
        buffer = BytesIO()
        image.save(buffer, format="JPEG")
        image_bytes = buffer.getvalue()

        # Create detailed prompt
        prompt = """
        Analyze the uploaded image carefully and generate
        a detailed description.

        Include:
        1. Main objects and subjects.
        2. Colors, shapes and appearance.
        3. Background and surroundings.
        4. Activities or actions taking place.
        5. Important details and visual elements.
        6. Overall context and meaning.

        Describe only what is visually observable.
        Do not make unsupported assumptions.

        Generate a clear, accurate and detailed description.
        """

        # Send image and prompt to Gemini VLM
        response = client.models.generate_content(
            model=MODEL,
            contents=[
                prompt,
                types.Part.from_bytes(
                    data=image_bytes,
                    mime_type="image/jpeg"
                )
            ]
        )

        return response.text

    except Exception as e:
        return f"Error generating description: {str(e)}"


print("Image Description Function Ready!")
```

```python
with gr.Blocks(theme=gr.themes.Soft()) as demo:

    gr.Markdown(
        """
        # AI Image Description Generator
        ### Multimodal AI using Vision-Language Model

        Upload an image and let AI generate
        a detailed description of its contents.
        """
    )

    with gr.Row():

        with gr.Column():
            input_image = gr.Image(
                label="Upload Image",
                type="pil"
            )

            generate_btn = gr.Button(
                "Generate Description",
                variant="primary"
            )

        with gr.Column():
            output_text = gr.Textbox(
                label="Generated Image Description",
                lines=15,
                interactive=False
            )

    generate_btn.click(
        fn=generate_description,
        inputs=input_image,
        outputs=output_text
    )

    gr.Markdown(
        "Powered by Gemini Vision-Language Model"
    )

demo.launch(share=True)
```
